```python
df.columns=df.columns.str.lower().str.replace(' ','_').str.replace('-','_')
```

```python
df['order_date']=pd.to_datetime(df['order_date'],dayfirst=True)
df['ship_date']=pd.to_datetime(df['ship_date'],dayfirst=True)
```

```python
df['profit_margin_pct'] = (df['profit'] / df['sales']) * 100
```

```python
df.columns.tolist()
```

```python
# 1. High-level Category performance
category_perf = df.groupby('category').agg({
    'sales': 'sum',
    'profit': 'sum',
    'order_id': 'count'
}).reset_index().sort_values(by='profit', ascending=False)

category_perf
```

```python
# 2. Granular Sub-Category performance (includes profit_margin_pct)
subcategory_perf = df.groupby(['category', 'sub_category']).agg({
    'sales': 'sum',
    'profit': 'sum',
    'profit_margin_pct': 'mean'
}).reset_index().sort_values(by='profit', ascending=False)

subcategory_perf
```

```python
# Define boundaries and readable bucket labels
bins = [-0.01, 0.0, 0.2, 0.5, 1.0]
labels = ['0% (No Discount)', '1% - 20% (Low)', '21% - 50% (Medium)', '> 50% (Heavy)']

# Create the new categorical column
df['discount_band'] = pd.cut(df['discount'], bins=bins, labels=labels)
```

```python
discount_impact = df.groupby('discount_band', observed=False).agg({
    'sales': 'sum',
    'profit': 'sum',
    'profit_margin_pct': 'mean',
    'order_id': 'count'
}).reset_index()

discount_impact
```

```python
# Aggregate by customer
customer_summary = df.groupby(['customer_id', 'customer_name']).agg({
    'sales': 'sum',
    'profit': 'sum',
    'order_id': 'nunique'  # Count distinct orders
}).reset_index()

# Top 10 most profitable customers
top_10_customers = customer_summary.nlargest(10, 'profit')
top_10_customers
```

```python
top_10_products_sales = df.groupby(['product_id', 'product_name', 'category']).agg({
    'sales': 'sum',
    'profit': 'sum',
    'quantity': 'sum'
}).reset_index().nlargest(10, 'sales')

top_10_products_sales
```

```python
top_10_loss_products = df.groupby(['product_id', 'product_name', 'sub_category']).agg({
    'profit': 'sum',
    'sales': 'sum',
    'discount': 'mean'
}).reset_index().nsmallest(10, 'profit')

top_10_loss_products
```

```python
import sqlite3

# 1. Create an in-memory SQL database connection
conn = sqlite3.connect(':memory:')

# 2. Write the cleaned DataFrame into an SQL table named 'superstore'
df.to_sql('superstore', conn, index=False, if_exists='replace')
```

```python
query = """
SELECT
    category,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY category
ORDER BY total_profit DESC;
"""

pd.read_sql(query, conn)
```

```python
query_1 = """
SELECT
    category,
    sub_category,
    ROUND(SUM(sales), 2) AS total_sales,
    ROUND(SUM(profit), 2) AS total_profit
FROM superstore
GROUP BY category, sub_category
HAVING total_sales > 500000 AND total_profit < 0
ORDER BY total_profit ASC;
"""

pd.read_sql(query_1, conn)
```

```python
# 1. Ensure order_date is datetime and extract order_year
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)
df['order_year'] = df['order_date'].dt.year

# 2. Refresh the SQLite table
df.to_sql('superstore', conn, index=False, if_exists='replace')

# 3. Verify the column exists in SQL
pd.read_sql("PRAGMA table_info(superstore);", conn)
```

```python
query_2 = """
SELECT
    market,
    order_year,
    ROUND(SUM(sales), 2) AS annual_sales,
    ROUND(SUM(profit), 2) AS annual_profit,
    COUNT(DISTINCT order_id) AS total_orders
FROM superstore
GROUP BY market, order_year
ORDER BY market ASC, order_year ASC;
"""

pd.read_sql(query_2, conn)
```
